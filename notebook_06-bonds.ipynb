# Obligation : pourquoi son prix baisse quand les taux montent

## 1. Question

> Une obligation promet toujours les mêmes flux. Pourtant son prix change tous les jours. Pourquoi baisse-t-il quand les taux montent ? Et que veut dire exactement « cette obligation rend 6 % » ?

## 2. Intuition

Une obligation, c'est un **prêt découpé en titres**. Vous prêtez 100 aujourd'hui, l'émetteur vous verse un **coupon** chaque année, puis vous rembourse le **nominal** à la fin. Les flux sont **fixes**, écrits à l'avance.

Mais le marché, lui, ne cesse de changer ce qu'il exige pour prêter. Si les taux montent, une nouvelle obligation paiera plus de coupon pour le même prix ; l'ancienne, qui paie moins, doit **baisser de prix** pour rester compétitive. Si les taux baissent, c'est l'inverse.

Le prix d'une obligation n'est donc que la **valeur actuelle de ses flux fixes**, actualisés au taux que le marché exige aujourd'hui. Vous avez déjà tout ce qu'il faut : c'est la vidéo sur l'actualisation, appliquée à un produit précis.

### Glossaire (nouveaux termes de cette vidéo)

| Terme | Définition simple |
|---|---|
| **Nominal / pair** (*par*, *face value*) | Montant remboursé à la fin, base de calcul du coupon. Ici 100. |
| **Taux de coupon** | Intérêt annuel promis, en % du nominal (5 % → 5 € par an). **Fixé à l'émission, il ne change plus.** |
| **Taux d'actualisation du marché / rendement** | Taux que le marché exige aujourd'hui pour prêter sur cette durée et ce risque. Il change tous les jours. |
| **Au pair / avec prime / avec décote** | Prix = 100 / > 100 / < 100. |
| **YTM** (*yield to maturity*, rendement actuariel) | Le taux unique qui, appliqué à **tous** les flux, redonne le prix de marché. On le trouve par calcul numérique. |
| **Taux spot (zéro-coupon)** | Taux d'un placement qui ne verse rien avant l'échéance (vidéo précédente). |
| **Rendement courant** (*current yield*) | Coupon annuel divisé par le prix. Ignore le temps et le gain ou la perte à l'échéance. |
| **Coupon couru** (*accrued interest*) | Part du prochain coupon déjà « gagnée » par le vendeur depuis le dernier paiement. |
| **Prix pied de coupon** (*clean price*, *flat price*) | Prix coté par les opérateurs, **sans** le coupon couru. |
| **Prix coupon couru inclus** (*dirty price*, *full price*) | Prix réellement payé = prix pied de coupon + coupon couru. |
| **Pull to par** | À l'approche de l'échéance, le prix converge vers 100 (si pas de défaut). |

## 3. Hypothèses

1. **Aucun risque de défaut** : tous les flux sont payés à la date prévue.
2. **Coupons annuels** dans la plupart des exemples (le cas semestriel est traité en 6.6).
3. **Un seul rendement** pour actualiser tous les flux (YTM), sauf quand on compare avec la courbe zéro-coupon (section 6.4).
4. **Calcul de prix à une date de coupon** (coupon couru nul), sauf en section 6.7.
5. **Pas de coûts de transaction, pas d'option intégrée** (pas d'obligation remboursable par anticipation).

## 4. Notation

| Symbole | Signification |
|---|---|
| $N$ | nombre de périodes de coupon restantes (années ici) |
| $PMT$ | coupon par période (5 si coupon 5 % sur 100) |
| $FV$ | nominal remboursé à l'échéance (100) |
| $y$ | rendement (YTM), taux d'actualisation du marché |
| $z_t$ | taux spot de maturité $t$ |
| $P$ | prix de l'obligation |
| $AI$ | coupon couru ; $t/T$ : fraction de la période de coupon écoulée |

## 5. Mathématiques (pas à pas)

### 5.1 Le prix d'une obligation
$$P=\sum_{k=1}^{N}\frac{PMT}{(1+y)^k}+\frac{FV}{(1+y)^N}$$

- $\frac{PMT}{(1+y)^k}$ : valeur actuelle du coupon versé à la fin de la période $k$.
- $\frac{FV}{(1+y)^N}$ : valeur actuelle du remboursement du nominal.
- $y$ : le **même** rendement pour tous les flux. C'est la simplification du YTM.
- Si $y$ augmente, chaque dénominateur augmente, donc chaque terme diminue : **le prix baisse**. C'est toute l'explication de la relation inverse prix-taux.

### 5.2 Au pair, avec prime, avec décote
Comparer le coupon $c$ et le rendement $y$ (CFA, *Fixed Income Analysis*) :
- $c = y$ → prix = 100 (**au pair**) ;
- $c < y$ → prix < 100 (**décote**) : le coupon est « insuffisant » ;
- $c > y$ → prix > 100 (**prime**) : le coupon est « excessif ».

L'écart de prix à 100 est exactement la **valeur actuelle de l'écart de coupon**. Exemple CFA : coupon 4 % contre rendement 6 %, 5 ans : on perd 2 par an pendant 5 ans, dont la valeur actuelle à 6 % vaut $-8{,}425$, d'où un prix de $100-8{,}425=91{,}575$.

### 5.3 Le YTM
Le YTM est la solution $y$ de
$$P_{marché}=\sum_{k=1}^{N}\frac{PMT}{(1+y)^k}+\frac{FV}{(1+y)^N}$$
Pas de formule fermée : on cherche $y$ numériquement. Ce n'est pas un taux observé, c'est un **taux déduit du prix**.

### 5.4 Prix avec les taux spot, et le YTM comme moyenne
$$P=\sum_{k=1}^{N}\frac{PMT}{(1+z_k)^k}+\frac{FV}{(1+z_N)^N}$$
Chaque flux a **son** taux spot. Comme l'obligation n'a qu'un prix, le YTM est une sorte de **moyenne pondérée des taux spot** : il est compris entre le plus petit et le plus grand (CFA).

### 5.5 Coupon couru, prix pied de coupon et coupon inclus
$$AI=\frac{t}{T}\times PMT \qquad\qquad P_{full}=P_{flat}+AI$$
- $t$ : jours écoulés depuis le dernier coupon ; $T$ : jours dans la période de coupon.
- On cote le prix **pied de coupon** pour que le prix ne grimpe pas artificiellement chaque jour (le coupon couru augmente tous les jours) et ne s'effondre pas le jour du paiement du coupon.
- Convention de comptage : *actual/actual* (obligations d'État) ou *30/360* (obligations d'entreprises).

### 5.6 Rendement courant
$$\text{rendement courant}=\frac{\text{coupon annuel}}{P_{flat}}$$
Simple, mais il ignore la valeur temps et le gain ou la perte à l'échéance.

## 6. Implémentation

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import brentq

plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["font.size"] = 12
```

### 6.1 La fonction de prix

On génère les flux (coupon chaque période, nominal en plus à la dernière), on les actualise avec le rendement `y`, on additionne. Le paramètre `m` est la fréquence des coupons par an (1 par défaut).

```python
def prix_obligation(coupon, maturite, y, nominal=100, m=1):
    """Prix à une date de coupon. coupon et y en décimal (0.05 = 5 %)."""
    n = int(round(maturite * m))
    t = np.arange(1, n + 1)
    cf = np.full(n, nominal * coupon / m, dtype=float)
    cf[-1] += nominal
    return float(np.sum(cf / (1 + y / m) ** t))

print(f"5 ans, coupon 5 %, rendement 5 % : {prix_obligation(0.05, 5, 0.05):.2f}")
print(f"5 ans, coupon 5 %, rendement 3 % : {prix_obligation(0.05, 5, 0.03):.2f}")
print(f"5 ans, coupon 5 %, rendement 6 % : {prix_obligation(0.05, 5, 0.06):.2f}")
```

**Lecture :** les flux sont **exactement les mêmes** dans les trois lignes (5, 5, 5, 5, 105). Seul le rendement change, et le prix passe de 109,16 à 100 puis 95,79. C'est la réponse à la question du début.

### 6.2 Vérification : les exemples publiés

Avant d'aller plus loin, on contrôle nos fonctions contre les exemples du CFA : coupons 4 %, 6 % et 8 % sur 5 ans avec un rendement de 6 %.

```python
for c in [0.04, 0.06, 0.08]:
    print(f"coupon {c:.0%}, rendement 6 % : {prix_obligation(c, 5, 0.06):.3f}")
```

**Lecture :** 91,575 (décote) ; 100,000 (au pair) ; 108,425 (prime). On retrouve exactement les valeurs du CFA, y compris l'écart symétrique de 8,425.

### 6.3 Le YTM : le chemin inverse (du prix au rendement)

Jusqu'ici, on part du rendement pour trouver le prix. Sur un desk, on part souvent du **prix de marché** et on cherche le rendement qui correspond. Il n'y a pas de formule, donc on utilise `brentq` : on lui donne une fonction (écart entre prix recalculé et prix cible) et un intervalle, il renvoie le rendement qui annule l'écart.

```python
def ytm(coupon, maturite, prix, nominal=100, m=1):
    f = lambda y: prix_obligation(coupon, maturite, y, nominal, m) - prix
    return brentq(f, -0.5, 1.0)

prix_marche = 95.79
print(f"Prix {prix_marche} -> YTM = {ytm(0.05, 5, prix_marche):.4%}")
print(f"Contrôle : prix à ce YTM = {prix_obligation(0.05, 5, ytm(0.05, 5, prix_marche)):.2f}")
```

**Lecture :** le prix de 95,79 correspond à un rendement de 6,00 % (5,9994 % exactement, car 95,79 est un prix arrondi). Le contrôle refait l'aller : on retrouve le prix. Prix et rendement sont deux façons de dire la même chose. Sur un desk, un opérateur peut coter l'un ou l'autre.

### 6.4 Le YTM face aux taux spot

Maintenant, on valorise avec **un taux spot par flux** (vidéo précédente), puis on calcule le YTM correspondant. Exemple du CFA : taux spot de 9 %, 10 %, 11 % à 1, 2, 3 ans ; obligation de 3 ans, coupon 5 %, nominal 100.

```python
def prix_spot(cf, z):
    t = np.arange(1, len(cf) + 1)
    return float(np.sum(np.array(cf) / (1 + np.array(z)) ** t))

def ytm_flux(cf, prix):
    t = np.arange(1, len(cf) + 1)
    return brentq(lambda y: np.sum(np.array(cf) / (1 + y) ** t) - prix, -0.5, 1.0)

cf3 = [5, 5, 105]
z3 = [0.09, 0.10, 0.11]
p3 = prix_spot(cf3, z3)
print(f"Prix avec les taux spot : {p3:.2f}")
print(f"YTM de cette obligation  : {ytm_flux(cf3, p3):.3%}")
print("Taux spot : 9,00 % / 10,00 % / 11,00 %")
```

**Lecture :** prix 85,49, YTM 10,93 %. Ce YTM est **compris entre 9 % et 11 %** : c'est une sorte de moyenne des trois taux spot. Plus proche de 11 % parce que le plus gros flux (105) tombe à l'année 3. Rappel de la vidéo précédente : trois obligations de coupons différents sur la même courbe donnent trois YTM différents. Le YTM résume, il ne cote pas.

### 6.5 Rendement courant, YTM : deux mesures qui ne disent pas la même chose

```python
coupon, mat, y_marche = 0.05, 5, 0.06
prix = prix_obligation(coupon, mat, y_marche)
rendement_courant = 100 * coupon / prix
print(f"Prix                : {prix:.2f}")
print(f"Rendement courant   : {rendement_courant:.3%}  (coupon / prix)")
print(f"YTM                 : {y_marche:.3%}")
```

**Lecture :** le rendement courant (5,22 %) est plus bas que le YTM (6 %), parce qu'il **ignore** que l'on achète sous 100 et que l'on sera remboursé à 100 : ce gain de 4,21 n'apparaît pas dans le coupon divisé par le prix. Le YTM, lui, le compte.

### 6.6 Convention de cotation : un rendement de 6 % annuel n'est pas un rendement de 6 % semestriel

Beaucoup d'obligations paient un coupon **semestriel**. Le rendement est alors coté en « base semestrielle » : on actualise chaque flux avec $y/2$ par semestre. Le même chiffre de 6 % ne donne donc pas le même prix.

```python
p_annuel = prix_obligation(0.05, 5, 0.06, m=1)
p_semestriel = prix_obligation(0.05, 5, 0.06, m=2)
print(f"Coupon annuel, rendement 6 %     : {p_annuel:.3f}")
print(f"Coupon semestriel, rendement 6 % : {p_semestriel:.3f}")
```

**Lecture :** les deux prix diffèrent légèrement. Un rendement sans sa convention de fréquence ne veut rien dire, comme un taux sans sa convention de composition dans la vidéo sur l'actualisation. Règle du CFA : composer plus souvent à un taux annuel plus faible équivaut à composer moins souvent à un taux plus élevé.

### 6.7 Prix pied de coupon et prix coupon inclus

Entre deux dates de coupon, le prix payé n'est pas le prix coté. On vérifie d'abord le calcul du coupon couru sur les exemples publiés.

```python
def coupon_couru(pmt, t_jours, T_jours):
    return pmt * t_jours / T_jours

# Hull : coupon 11 % semestriel -> 5,50 ; 54 jours sur 181 (actual/actual)
ai_hull = coupon_couru(5.50, 54, 181)
print(f"Hull : coupon couru = {ai_hull:.4f} ; prix coté 95,50 -> prix payé = {95.50 + ai_hull:.2f}")

# CFA : coupon 4,375 % semestriel, 43 jours sur 184 (actual/actual) vs 42 sur 180 (30/360)
pmt_cfa = 4.375 / 2
print(f"CFA actual/actual : {coupon_couru(pmt_cfa, 43, 184):.6f}")
print(f"CFA 30/360        : {coupon_couru(pmt_cfa, 42, 180):.6f}")
```

**Lecture :** on retrouve 1,64 (prix payé 97,14) chez Hull, et 0,511209 / 0,510417 chez le CFA. Le choix de la convention de comptage de jours change légèrement le coupon couru.

On construit maintenant le **prix coupon inclus** pour toute date, à rendement constant : les flux sont actualisés avec des exposants décalés de la fraction de période déjà écoulée.

```python
def prix_plein(pmt, N, y, f, nominal=100):
    """Prix coupon inclus. N coupons restants, f = fraction de période écoulée."""
    k = np.arange(1, N + 1)
    cf = np.full(N, float(pmt)); cf[-1] += nominal
    return float(np.sum(cf / (1 + y) ** (k - f)))

# Contrôle : f = 0 doit redonner le prix de la section 6.1
print(f"f = 0 : {prix_plein(5, 5, 0.06, 0):.2f}  (attendu {prix_obligation(0.05, 5, 0.06):.2f})")

# Une obligation de 3 ans, coupon 5 %, rendement constant 6 % : on avance dans le temps
tau = np.linspace(0, 2.995, 600)
plein, pied = [], []
for x in tau:
    k = int(np.floor(x)); f = x - k
    full = prix_plein(5, 3 - k, 0.06, f)
    plein.append(full); pied.append(full - 5 * f)
print(f"Début : plein = {plein[0]:.2f}, pied de coupon = {pied[0]:.2f}")
```

## 7. Visualisation

**Graphique 1 — Quelle est la relation entre rendement et prix ?**

```python
ys = np.linspace(0.0, 0.12, 120)
prix_y = [prix_obligation(0.05, 5, y) for y in ys]

fig, ax = plt.subplots()
ax.plot(ys * 100, prix_y)
for y in [0.03, 0.06]:
    p = prix_obligation(0.05, 5, y)
    ax.plot(y * 100, p, "o"); ax.annotate(f"{p:.2f}", (y * 100, p), textcoords="offset points", xytext=(8, 6))
ax.axhline(100, color="grey", ls="--", lw=1)
ax.set_xlabel("Rendement (%)"); ax.set_ylabel("Prix (pour 100 de nominal)")
ax.set_title("Mêmes flux, rendements différents : la courbe prix-rendement")
plt.show()
```

**Lecture :** la courbe est **décroissante** (prix et taux vont en sens inverse) et **courbée** : elle n'est pas une droite. À 5 % (rendement = coupon) le prix vaut exactement 100. La courbure sera au cœur de la vidéo sur la convexité.

**Graphique 2 — Quelle obligation réagit le plus à une hausse des taux ?**

```python
maturites = [2, 5, 10, 30]
coupons = [0.02, 0.05, 0.08]
largeur = 0.25
fig, ax = plt.subplots()
for i, c in enumerate(coupons):
    var = []
    for N in maturites:
        p0 = prix_obligation(c, N, 0.05); p1 = prix_obligation(c, N, 0.06)
        var.append((p1 / p0 - 1) * 100)
    ax.bar(np.arange(len(maturites)) + i * largeur, var, largeur, label=f"coupon {c:.0%}")
ax.set_xticks(np.arange(len(maturites)) + largeur); ax.set_xticklabels([f"{N} ans" for N in maturites])
ax.set_ylabel("Variation du prix pour +1 % de rendement (%)")
ax.set_title("Plus la maturité est longue, plus le coupon est bas : plus ça bouge")
ax.legend(); plt.show()
```

**Lecture :** pour une hausse de rendement de 5 % à 6 %, une obligation à 2 ans perd environ 1,8 % ; une à 30 ans, de 13 % à 17 %. Et à maturité égale, le coupon le plus bas réagit le plus. Ce sont les deux facteurs que la **duration** (prochaine vidéo) va résumer en un seul chiffre.

**Graphique 3 — Que devient le prix avec le temps, si les taux ne bougent pas ?**

```python
annees = np.arange(5, -1, -1)
fig, ax = plt.subplots()
for c, nom in [(0.08, "coupon 8 % (prime)"), (0.04, "coupon 4 % (décote)")]:
    prix_t = [prix_obligation(c, N, 0.06) if N > 0 else 100.0 for N in annees]
    ax.plot(5 - annees, prix_t, "o-", label=nom)
ax.axhline(100, color="grey", ls="--", lw=1)
ax.set_xlabel("Années écoulées depuis l'achat"); ax.set_ylabel("Prix")
ax.set_title("Pull to par : à rendement constant, le prix converge vers 100")
ax.legend(); plt.show()
```

**Lecture :** sans aucun mouvement de taux, le prix d'une obligation avec prime baisse jusqu'à 100, celui d'une obligation avec décote monte jusqu'à 100. Il ne faut donc pas confondre « le prix change » et « le marché a bougé ».

**Graphique 4 — Pourquoi coter le prix pied de coupon ?**

```python
fig, ax = plt.subplots()
ax.plot(tau, plein, label="prix coupon inclus (payé)")
ax.plot(tau, pied, label="prix pied de coupon (coté)")
ax.set_xlabel("Années écoulées"); ax.set_ylabel("Prix")
ax.set_title("Même rendement tout du long : le prix payé fait des dents de scie")
ax.legend(); plt.show()
```

**Lecture :** le prix coupon inclus monte chaque jour (le coupon couru augmente) puis chute le jour du paiement du coupon : une courbe en dents de scie qui ne reflète aucun mouvement de marché. Le prix pied de coupon est beaucoup plus lisse : c'est pour cela qu'il est coté.

## 8. Expériences

### 8.1 Que garantit vraiment le YTM ?

Le YTM suppose que les coupons sont **réinvestis au YTM**. Si les taux ont baissé entre-temps, ce n'est pas le cas. On compare, pour l'obligation 5 ans à coupon 5 % achetée à 95,79 (YTM 6 %), le rendement réalisé selon le taux de réinvestissement des coupons.

```python
prix = prix_obligation(0.05, 5, 0.06)

def rendement_realise(prix, taux_reinvest, coupon=5, N=5, nominal=100):
    valeur_finale = sum(coupon * (1 + taux_reinvest) ** (N - k) for k in range(1, N + 1)) + nominal
    return (valeur_finale / prix) ** (1 / N) - 1

for r in [0.06, 0.03]:
    print(f"Coupons réinvestis à {r:.0%} : rendement réalisé = {rendement_realise(prix, r):.3%}")
```

**Lecture :** si les coupons sont réinvestis à 6 %, on retrouve bien le YTM de 6,00 %. S'ils sont réinvestis à 3 %, le rendement réalisé n'est que de 5,73 %. Le YTM est une **promesse conditionnelle**, pas un rendement garanti. Le CFA rappelle d'ailleurs que le YTM est un mauvais estimateur si les taux sont volatils ou si la courbe n'est pas plate.

### 8.2 Un choc de rendement, deux obligations

Reprenons deux obligations de maturité 10 ans, l'une à coupon 2 %, l'autre à 8 %, au même rendement de 5 %.

```python
for c in [0.02, 0.08]:
    p0 = prix_obligation(c, 10, 0.05)
    p_up = prix_obligation(c, 10, 0.06)
    p_dn = prix_obligation(c, 10, 0.04)
    print(f"coupon {c:.0%} : prix {p0:6.2f} | +1 % : {(p_up/p0-1):+.2%} | -1 % : {(p_dn/p0-1):+.2%}")
```

**Lecture :** la hausse de 1 % fait perdre 8,17 % à l'obligation à 2 % et 6,86 % à celle à 8 %. Et regardez l'asymétrie : la baisse de 1 % fait **gagner plus** (+9,04 %) que la hausse de 1 % ne fait perdre. C'est la **convexité**, qu'on étudiera dans deux vidéos.

## 9. Résultats

- À flux inchangés (5, 5, 5, 5, 105), le prix passe de **109,16 à 95,79** quand le rendement passe de 3 % à 6 %.
- Au pair si coupon = rendement ; décote si coupon < rendement ; prime si coupon > rendement. Les valeurs du CFA sont retrouvées (91,575 ; 100 ; 108,425).
- Le YTM est un rendement **déduit du prix**, et une sorte de moyenne des taux spot (10,93 % entre 9 % et 11 % dans l'exemple du CFA).
- Le rendement courant (5,22 %) sous-estime le YTM (6 %) parce qu'il ignore le gain à l'échéance.
- Le prix converge vers 100 avec le temps (*pull to par*), sans qu'aucun taux ne bouge.
- Le prix coté est le prix pied de coupon ; le prix payé ajoute le coupon couru (1,64 chez Hull, exemple publié).
- Les coupons réinvestis à 3 % au lieu de 6 % ramènent le rendement réalisé à 5,73 %.

## 10. Interprétation financière

- **Trader taux :** il parle en prix et en rendement, selon le marché. La relation inverse prix-taux est la base de son risque : tout son book bouge quand la courbe bouge.
- **Gérant obligataire :** il achète un YTM, mais son rendement réalisé dépend du réinvestissement et du fait de garder jusqu'à l'échéance.
- **Risk manager :** deux obligations de même rendement peuvent réagir très différemment à un choc de taux (maturité, coupon). C'est le point de départ des mesures de sensibilité (duration, DV01).
- **Back-office / opérations :** la différence entre prix coté et prix payé (coupon couru) est un classique des erreurs de règlement.

## 11. Limites

- **Pas de risque de défaut** : une obligation d'entreprise s'actualise avec un spread de crédit.
- **Un seul rendement pour tous les flux** : le YTM résume, il ne cote pas ; la courbe zéro est la référence.
- **Réinvestissement au YTM** : hypothèse forte, rarement vérifiée.
- **Dates de coupon et jours ouvrés** : les conventions réelles (actual/actual, 30/360, jours fériés) sont plus riches que nos exemples.
- **Pas d'option intégrée** (remboursement anticipé, conversion) : le YTM devient alors un mauvais outil (*yield to worst*).
- **Taux et courbes construits pour la pédagogie**, sauf les exemples publiés cités.

## 12. Conclusion

> **Les flux d'une obligation sont fixes ; son prix est la valeur actuelle de ces flux au taux que le marché exige aujourd'hui. Quand le taux monte, le prix baisse : c'est de l'arithmétique, pas un mystère.**

**Prochaine vidéo :** de combien baisse-t-il ? C'est la duration.
